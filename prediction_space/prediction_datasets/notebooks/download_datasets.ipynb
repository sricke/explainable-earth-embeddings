## Biome — RESOLVE Ecoregions 2017 (14 biomes)

Biome and Ecoregion share the same shapefile — downloading one skips the download for the other.

```python
ds_biome = BiomeDataset()
ds_biome.download()
ds_biome.verify_download()
```

## Ecoregion — RESOLVE Ecoregions 2017 (~800 regions)

```python
ds_eco = EcoregionDataset()
ds_eco.verify_download()  # no download needed if biome already ran
```

## NDVI — MODIS MOD13C2 (requires NASA Earthdata credentials)

Register at https://urs.earthdata.nasa.gov/ and set credentials below.

```python
import os

username = os.environ.get("EARTHDATA_USERNAME")
password = os.environ.get("EARTHDATA_PASSWORD")
```

```python
if username and password:
    ds_ndvi = NDVIDataset()
    ds_ndvi.download(username=username, password=password)
    ds_ndvi.verify_download()
else:
    print("Skipping NDVI — set EARTHDATA_USERNAME and EARTHDATA_PASSWORD above.")
```

## Nightlights — VIIRS DNB Annual Composite (manual download)

1. Register (free) at https://eogdata.mines.edu/products/vnl/
2. Download the annual `average_masked` GeoTIFF
3. Place it at `~/data/prediction_tasks/VNL_annual_average_masked.tif`  
   (or pass `file_path=` when constructing `NightlightsDataset`)

```python
ds_nl = NightlightsDataset()
ds_nl.verify_download()
```

## Summary

```python
datasets = {
    "Elevation":         ds_elev,
    "Temperature":       ds_temp,
    "Precipitation":     ds_prec,
    "Population":        ds_pop,
    "Soil Carbon":       ds_soc,
    "Köppen Climate":    ds_kgc,
    #"Land Cover":        ds_lc,
    "Biome":             ds_biome,
    "Ecoregion":         ds_eco,
}

print(f"{'Dataset':<20} {'Status':<10} {'N pixels':>12}")
print("-" * 45)
for name, ds in datasets.items():
    ok = ds.verify_download()
    if ok:
        try:
            n = len(ds)
            print(f"{name:<20} {'OK':<10} {n:>12,}")
        except Exception as e:
            print(f"{name:<20} {'ERROR':<10} {str(e)[:20]}")
    else:
        print(f"{name:<20} {'MISSING':<10} {'—':>12}")
```
